## Project Summary -

This project focuses on analyzing the stock price movements of major technology companies using real-world financial data. The primary objective is to extract meaningful insights from stock data by applying data analytics techniques such as data cleaning, feature engineering, and visualization.

The project also includes advanced visualizations such as moving average crossover analysis and cumulative return graphs, which provide deeper insights into market behavior and investment potential.

Overall, this project demonstrates how data analytics can be applied to financial datasets to understand trends, measure risk, and compare company performance. ### Variable Description

The dataset consists of the following variables:

* **stock_symbol**: Represents the company’s stock ticker symbol (e.g., AAPL for Apple, MSFT for Microsoft).

* **date**: Indicates the trading date for which the stock data is recorded.

* **open**: The price at which the stock started trading at the beginning of the day.

* **high**: The highest price reached by the stock during the trading day.

* **low**: The lowest price recorded during the trading day.

* **close**: The final price of the stock at the end of the trading session.

* **adj_close**: The adjusted closing price that accounts for corporate actions like dividends and stock splits.

* **volume**: The total number of shares traded during the day, indicating market activity.



```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df = df.sort_values(by='date')
```

```python
df.isnull().sum()
```

```python
df.fillna(method='ffill', inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df['price_change'] = df['close'] - df['open']
```

```python
df['price_range'] = df['high'] - df['low']
```

```python
df['pct_change'] = ((df['close'] - df['open']) / df['open']) * 100
```

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
```

```python
df.head()
```

```python
apple = df[df['stock_symbol']=='AAPL']
```

```python
apple['MA_7'] = apple['close'].rolling(7).mean()
apple['MA_30'] = apple['close'].rolling(30).mean()
```

```python
plt.figure()

plt.plot(apple['date'], apple['close'], label='Close Price')
plt.plot(apple['date'], apple['MA_7'], label='7-Day MA')
plt.plot(apple['date'], apple['MA_30'], label='30-Day MA')

plt.legend()
plt.title("Apple Stock Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)

plt.show()
```

### Why this graph?
This graph is used to analyze the overall stock price trend of Apple over time along with moving averages. The 7-day and 30-day moving averages help smooth short-term fluctuations and reveal both short-term and long-term trends.

### Insights from the graph
The graph shows a clear long-term upward trend in Apple’s stock price, indicating strong growth over the years. The 7-day moving average closely follows the price, capturing short-term variations, while the 30-day moving average provides a smoother long-term trend. Crossovers between the moving averages indicate possible trend reversals, and deviations from the averages highlight periods of volatility.

```python
plt.figure()

plt.plot(apple['date'], apple['price_range'])

plt.title("Daily Price Range (Volatility) - Apple")
plt.xlabel("Date")
plt.ylabel("Price Range")
plt.xticks(rotation=45)

plt.show()
```

### Why this graph?

```python
apple['volatility'] = apple['close'].rolling(7).std()
plt.figure()

plt.plot(apple['date'], apple['volatility'])

plt.title("7-Day Rolling Volatility (Apple)")
plt.xlabel("Date")
plt.ylabel("Volatility")
plt.xticks(rotation=45)

plt.show()
```

### Why this graph?
This graph is used to measure the rolling volatility of Apple’s stock using a 7-day window. It calculates the standard deviation of closing prices over time, helping to understand how unstable or risky the stock is during different periods.

### Insights from the graph
The graph shows that volatility was relatively low and stable in earlier years, indicating consistent stock behavior. However, after around 2019, volatility increases significantly with frequent spikes, suggesting higher uncertainty and risk in the market. These spikes represent periods where stock prices changed rapidly within a short time frame.

```python
plt.figure()

plt.plot(apple['date'], apple['MA_7'], label='7-Day MA')
plt.plot(apple['date'], apple['MA_30'], label='30-Day MA')

plt.title("Moving Average Crossover (Apple)")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

### Why this graph?
This graph is used to analyze the interaction between short-term and long-term moving averages. It helps in identifying trend changes and potential buy or sell signals in stock price movements.

### Insights from the graph
The graph shows that the 7-day moving average closely follows short-term price movements, while the 30-day moving average represents the long-term trend. Points where the 7-day moving average crosses above the 30-day moving average indicate potential upward trends (buy signals), while crossovers below suggest possible downward trends (sell signals). The graph highlights periods where trend reversals occur, especially during phases of rapid price movement.

```python
apple['volume_MA_7'] = apple['volume'].rolling(7).mean()

plt.figure()

plt.plot(apple['date'], apple['volume'], alpha=0.3, label='Daily Volume')
plt.plot(apple['date'], apple['volume_MA_7'], label='7-Day Avg Volume')

plt.title("Rolling Volume Trend (Apple)")
plt.xlabel("Date")
plt.ylabel("Volume")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

### Why this graph?
This graph is used to analyze the trend in trading volume by comparing daily volume with its 7-day moving average. It helps in smoothing short-term fluctuations and identifying overall patterns in market activity.

### Insights from the graph
The graph shows that daily trading volume is highly volatile, with frequent spikes in earlier years. However, the 7-day moving average line smooths these fluctuations and reveals a clear declining trend over time. This indicates that trading activity was higher in earlier years and gradually decreased, stabilizing at lower levels. Occasional spikes still occur, representing temporary increases in market activity.

```python
apple['cumulative_return'] = (apple['close'] / apple['close'].iloc[0]) * 100

plt.figure()

plt.plot(apple['date'], apple['cumulative_return'])

plt.title("Cumulative Return of Apple Stock (Base = 100)")
plt.xlabel("Date")
plt.ylabel("Cumulative Return")

plt.xticks(rotation=45)
plt.show()
```

### Why this graph?
This graph is used to measure the overall growth of Apple’s stock over time by converting price into a cumulative return index. It helps in understanding how an initial investment would have grown.

### Insights from the graph
The graph shows a significant increase in cumulative returns over time, indicating strong long-term growth. The steep rise in later years highlights accelerated growth, especially after 2019. This demonstrates that long-term investment in the stock would have yielded substantial returns despite short-term fluctuations.
